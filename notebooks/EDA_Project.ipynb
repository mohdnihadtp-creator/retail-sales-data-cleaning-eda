```python
import pandas as pd
df=pd.read_csv('/content/retail_sales_raw.csv')
df.head()
```

```python
for i in df.columns:
  print(i)
  print(df[i].unique())
```

```python
for col in df.select_dtypes(include='object').columns:
    df[col]=df[col].str.strip()
```

```python
for col in df.select_dtypes(include='object').columns:
    df[col]=df[col].str.replace('%','', regex=True)
```

```python
df['Discount'].unique()
```

```python
df.columns
```

```python
df.columns=df.columns.str.lower().str.replace(' ','_')
```

```python
df.dtypes
```

```python
df[['discount','sales']].head(10)
```

```python
df['discount']=pd.to_numeric(df['discount'],errors='coerce')
df['sales']=pd.to_numeric(df['sales'],errors='coerce')
df['category'] = df['category'].str.strip().str.title()
```

```python
df[['discount','sales']].dtypes
```

```python
df[['order_date', 'ship_date']].dtypes
```

```python
df['order_date']=pd.to_datetime(
    df['order_date'],
    errors='coerce')
df['ship_date']=pd.to_datetime(
    df['ship_date'],
    errors='coerce')
```

```python
df[['order_date', 'ship_date']].dtypes
```

```python
df.isnull().sum()
```

```python
missing_values=df.isnull().sum()
missing_values=missing_values[missing_values > 0].sort_values(ascending=False)
print(missing_values)
```

```python
categorical_cols = [
    'customer_name',
    'customer_city',
    'payment_method',
    'sub-category'
]
for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
df['discount']=df['discount'].fillna(0)
df['quantity']=df['quantity'].fillna(df['quantity'].median())
df['unit_cost']=df.groupby('product_name')['unit_cost'].transform(lambda x:x.fillna(x.median()))
df['sales']=df['sales'].fillna(df['quantity']*df['unit_selling_price'])
df['profit']=df['profit'].fillna( df['sales']-(df['quantity']*df['unit_cost'])-df['shipping_cost']+df['other_cost_adjustment'])
df['ship_date']=df['ship_date'].fillna(df['ship_date'].mode()[0])
df['order_date']=df['order_date'].fillna(df['order_date'].mode()[0])
```

```python
df.isnull().sum()
```

```python
print('Duplicates :',df.duplicated().sum())
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
numeric_df=df.select_dtypes(include='number')
numeric_df=numeric_df.drop(columns=['row_id'],errors='ignore')
correlation=numeric_df.corr()
correlation
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(12, 8))
sns.heatmap(correlation,annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation Heatmap - Retail Sales')
plt.savefig("heatmap.png",dpi=300,bbox_inches="tight")
plt.show()

from google.colab import files
files.download("heatmap.png")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.barplot(data=df,x='category',y='sales',estimator=sum)
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.savefig("Sales_by_Category.png",dpi=300,bbox_inches="tight")
plt.show()

from google.colab import files
files.download("Sales_by_Category.png")
```

This bar chart compares total sales across different product categories. It helps identify which categories contribute the most and least to overall sales.


```python
plt.figure(figsize=(8,5))
sns.barplot(data=df,x='customer_segment',y='sales',estimator=sum,hue='customer_segment',palette=['yellow','green','red'],legend=False)
plt.title('Total Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales')
plt.show()
```

This chart compares sales across customer segments. It helps us understand which customer groups contribute most to the business.


```python
plt.figure(figsize=(8,5))
sns.barplot(data=df,x='category',y='profit',estimator=sum)
plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.show()
```

This chart compares the total profit generated by each product category. It helps identify the most profitable categories and those that may need improvement.


```python
df['month']=df['order_date'].dt.to_period('M')
monthly_sales=df.groupby('month')['sales'].sum()
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index.astype(str),monthly_sales.values)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid()
plt.savefig("mothly_sales.png",dpi=300,bbox_inches="tight")
plt.show()

from google.colab import files
files.download("mothly_sales.png")
```

I used a line chart to analyze monthly sales trends. It helps identify changes in sales over time and the months with higher or lower sales.


```python
top_products=df.groupby('product_name')['sales'].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
top_products.sort_values().plot(kind='barh')
plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.savefig("product_by_sales.png",dpi=300,bbox_inches="tight")
plt.show()
from google.colab import files
files.download('product_by_sales.png')
```

This bar chart displays the top ten products based on total sales. It helps identify the products that contribute most to sales revenue.

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8,5))
sns.scatterplot(data=df,x='sales',y='profit')
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

This scatter plot shows the relationship between sales and profit. Each point represents a record in the dataset. It helps us understand whether records with higher sales also tend to have higher profits.

```python
df['customer_segment']=df['customer_segment'].str.strip().str.title()
```

I standardized the customer segment names to avoid duplicate categories caused by differences in capitalization, such as "Consumer" and "consumer".

```python
sales_by_segment = df.groupby('customer_segment')['sales'].sum()
plt.figure(figsize=(8, 5))
plt.pie(
    sales_by_segment,
    autopct='%1.1f%%',
    startangle=90)
plt.legend(
    sales_by_segment.index,
    title='Customer Segment',
    loc='center left',
    bbox_to_anchor=(1, 0.5))
plt.title('Sales by Customer Segment')
plt.tight_layout()
plt.show()
```

This pie chart shows the percentage contribution of each customer segment to total sales. It helps us identify which segment contributes the most to sales.
